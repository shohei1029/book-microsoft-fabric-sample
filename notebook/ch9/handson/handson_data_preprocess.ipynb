```python
from pyspark.sql.types import *

schema = StructType([
    StructField("Store", IntegerType(), True),
    StructField("Date", DateType(), True),
    StructField("Weekly_Sales", DoubleType(), True),
    StructField("Holiday_Flag", IntegerType(), True),
    StructField("Temperature", DoubleType(), True),
    StructField("Fuel_Price", DoubleType(), True),
    StructField("CPI", DoubleType(), True),
    StructField("Unemployment", DoubleType(), True),
])

df = (spark.read.format("csv")
      .option("header", "true")
      .option("nullValue", "")                 
      .option("dateFormat", "dd-MM-yyyy")      
      .schema(schema)                          
      .load("Files/Walmart.csv"))

display(df)
```

```python
from pyspark.sql.functions import col

df = df.withColumn("Holiday_Flag", (col("Holiday_Flag") == 1).cast("boolean"))
display(df)
```

```python
train_df, test_df = df.randomSplit([0.9, 0.1], seed=42)

train_df.write.mode("overwrite").format("delta").saveAsTable("walmart_sales_dataset_train")
test_df.write.mode("overwrite").format("delta").saveAsTable("walmart_sales_dataset_test")
```
