```python
from synapse.ml.predict import MLFlowTransformer

df_test = spark.read.format("delta").load("Tables/dbo/walmart_sales_dataset_test")

model = MLFlowTransformer(
    inputCols=["Date", "Store", "Temperature", "Fuel_Price", "CPI", "Unemployment", "Holiday_Flag"],
    outputCol='predictions',
    modelName='handson_automl-AutoMLModel',
    modelVersion=1
)

predictions = model.transform(df_test)
display(predictions)
```

```python
predictions.write.mode("overwrite").format("delta").saveAsTable("walmart_sales_dataset_predictions")
```
