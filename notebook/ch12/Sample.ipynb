```python
# Key Fix: The extract method returns multiple columns (one for each label). To add them to the DataFrame correctly, join them back into the DataFrame.
# - Instead of assigning the result to a single column, pd.concat is used to join the result to the original DataFrame.
# - Each extracted entity gets its own column.

import pandas as pd

# Example Japanese entity phrases
df = pd.DataFrame([
    "田中太郎は東京の本社で、プロジェクトAの担当をしています。",
    "鈴木花子は2023年に大阪支社からロンドンオフィスへ異動しました。",
], columns=["text"])

# Extract entities: 人名 (person), 地名 (location), プロジェクト名 (project)
df_extracted = df["text"].ai.extract("人名", "地名", "プロジェクト名")

display(df_extracted)
```

```python
# FAQの問い合わせデータと既存ナレッジ記事のタイトルを比較し、
# 「最も近いナレッジ記事」を自動マッチングするユースケース
df = pd.DataFrame([
    ("ログインしようとするとパスワードエラーになります",
     "パスワードを忘れた場合の再設定手順"),
    ("請求書のPDFがダウンロードできません",
     "請求書ダウンロード時のトラブルシューティング"),
    ("解約方法を教えてください",
     "新規アカウントの作成方法"),
    ("二段階認証のコードが届かない",
     "二段階認証コードが受信できないときの対処法"),
], columns=["inquiry", "kb_article_title"])

# 1) 単一参照テキストとの類似度：問い合わせが「解約・退会」に関するものか判定
df["churn_intent_score"] = df["inquiry"].ai.similarity(
    "サービスの解約・退会を希望する問い合わせ"
)

# 2) 列同士のペアワイズ比較：問い合わせと候補ナレッジ記事のマッチ度を算出
df["kb_match_score"] = df["inquiry"].ai.similarity(df["kb_article_title"])

display(df)
```

```python
df = pd.DataFrame([
    ("ノートPC", "バッテリーが12時間持続し、重量わずか1.2kg。"),
    ("スマートウォッチ", "心拍数・睡眠・活動量をリアルタイムで計測。"),
], columns=["product", "feature"])

# テンプレート形式のプロンプトで各行に合わせた文章を生成
df["marketing_copy"] = df.ai.generate_response(
    prompt="{product}の特徴「{feature}」を活かした、魅力的なマーケティングコピーを1文で作成してください。",
    is_prompt_template=True
)
display(df)
```

```python
folder_path = "/lakehouse/default/Files/images"
file_paths = aifunc.list_file_paths(folder_path)
df = pd.DataFrame({"file_path": file_paths}).reset_index(drop=True)

# 製品名を付与
product_names = {"productA": "製品A", "productB": "製品B", "productC": "製品C"}
df["product"] = df["file_path"].astype(str).apply(
    lambda p: next((v for k, v in product_names.items() if k in p), "不明")
)
display(df)
```

```python
df["image_description"] = df["file_path"].ai.generate_response(
    prompt="この画像に写っている主要なオブジェクト、色、レイアウト、テキストを含む自然言語の説明文を日本語で生成してください。",
    column_type="path",
)
display(df[["product", "image_description"]])
```

```python
# 前述の ai.describe_image で生成した df（product, file_path, image_description）を流用
# 製品A〜Cの画像から色（赤／青／緑／白／黒／その他）を自動分類するパイプライン

# Step 1: キャプション（image_description）から埋め込みベクトルを生成
df["embedding"] = df["image_description"].ai.embed()
# DoubleType エラーを防ぐため float に明示変換
df["embedding"] = df["embedding"].apply(lambda v: [float(x) for x in v])

# Step 2: キャプションをもとに代表色で自動分類
df["color_category"] = df["image_description"].ai.classify(
    "赤色", "青色", "緑色", "白色", "黒色", "その他"
)

display(df[["product", "image_description", "color_category"]])
```
